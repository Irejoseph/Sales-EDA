Before any grouping occurs, I'll like to answer; What are the total amount?, total profit? and total quantity? represented by all 1194 records.

```python
total_amount = rawsales["Amount"].sum()

total_profit = rawsales["Profit"].sum()

total_quantity = rawsales["Quantity"].sum()
```

```python
print("Total Amount:", total_amount)
print("Total Profit:", total_profit)
print("Total Quantity:", total_quantity)
```

Total amount = 6,182,639.

Total profit = 1,610,697.

Total quantity = 12,745.

Bear in mind 1,194 rows ≠ 1,194 orders

The dataset records a total Amount of approximately 6.18 million, total Profit of approximately 1.61 million, and a total Quantity of 12,745 units across 1,194 records.

```python
# Calculate the profit margin percentage using the formula:
profit_margin = (total_profit / total_amount) * 100
```

what proportion of the total amount is represented by profit, then convert that proportion to a percentage by multiplying by 100. 
This gives you the profit margin percentage, which indicates how much profit is generated for every dollar of sales.

```python
print(round(profit_margin, 2))
```

Print the profit margin percentage rounded to two decimal places.

So, if Amount represents the sales base against which Profit is measured, the dataset has an overall profit-to-amount ratio of approximately 26.05%.

Average Amount by Category

```python
rawsales.groupby("Category")["Amount"].mean()
```

Furniture records have a lower average Amount (5,009) than Electronics (5,295) and Office Supplies (5,237). we already discovered that a row doesn't necessarily equal one identifiable sale/order given the weirdness with Order_ID.

Total and average Amounts by Sub-category

```python
total_subcat = rawsales.groupby("Sub-Category")["Amount"].sum().sort_values(ascending=False)

average_subcat = rawsales.groupby("Sub-Category")["Amount"].mean().sort_values(ascending=False)
```

```python
print("Total Amount by Sub-Category: ", [total_subcat])
```

```python
print("Average Amount by Sub-Category:", [average_subcat])
```

Markers generated the highest total Amount (627,875), closely followed by Tables (625,177). However, Printers recorded the highest average Amount per record (5,961.67), despite ranking fourth by total Amount. This suggests that total performance is influenced not only by the value of individual records but also by how frequently each sub-category occurs

Are Markers and Tables generating high total Amount partly because they appear more frequently in the dataset?

```python
rawsales["Sub-Category"].unique()
```

```python
Record_Count = rawsales["Sub-Category"].value_counts()
```

```python
subcat_comparison = pd.concat(
    [Record_Count, total_subcat, average_subcat], axis=1
)
```

```python
subcat_comparison.columns = ["Record_Count", "Total_Amount", "Average_Amount"]
```

```python
subcat_comparison
```

Sub-categories with more records generally tend to produce higher Total Amounts. However, record frequency alone does not determine total performance; Average Amount also plays an important role. Markers combine a relatively high record count with a high Average Amount, resulting in the highest Total Amount, while Printers demonstrate that a high Average Amount can compensate for a lower record count.

```python
subcat_comparison["Average_Amount"] = subcat_comparison["Average_Amount"].round(2)
```

```python
subcat_comparison
```

Analysis update

The dataset contains 1,194 records covering product transactions across three main categories: Electronics, Furniture, and Office Supplies.

Initial data-quality checks found no missing values and no exact duplicate rows. The Order_Date column was converted from an object to a datetime datatype. The existing Year-Month column was verified against the year-month derived from Order_Date, with all 1,194 records matching, making the column redundant for the analysis.

Investigation of Order_ID found 547 unique IDs across 1,194 records. Individual IDs can occur across different customers, dates and locations, meaning Order_ID should not be treated as a reliable unique identifier for individual transactions without further documentation.

For overall performance, the dataset records a Total Amount of 6,182,639, Total Profit of 1,610,697, and Total Quantity of 12,745. Profit represents approximately 26.05% of the total Amount, assuming these fields are intended to be compared on that basis.

At category level, total Amount is relatively evenly distributed:

Office Supplies    2,089,510

Electronics        2,054,456

Furniture          2,038,673

Despite Furniture having the highest number of records (407), it has the lowest total Amount. Its average Amount per record (5,009.02) is also lower than Electronics (5,294.99) and Office Supplies (5,236.87), helping explain this difference.

At sub-category level, Markers generated the highest Total Amount (627,875), narrowly followed by Tables (625,177). However, Printers had the highest Average Amount per record (5,961.67) despite ranking fourth in Total Amount.

Comparing record frequency with average and total Amount suggests that both volume and value per record contribute to overall sub-category performance. Tables benefit from having the highest record count (122), while Printers demonstrate that a high Average Amount can compensate for fewer records (95).
