The dataset contains 1,194 records covering sales transactions across three main product categories: Electronics, Furniture, and Office Supplies.

Initial data-quality checks identified no missing values and no exact duplicate rows. The Order_Date column was converted from an object to a datetime datatype. The existing Year-Month column was also checked against the year and month derived directly from Order_Date, with all 1,194 records matching, confirming that the existing column was consistent but redundant for the analysis.

Investigation of Order_ID identified 547 unique IDs across 1,194 records. The same Order IDs can occur across different customers, dates and locations, meaning Order_ID should not be assumed to uniquely identify individual transactions without further documentation about the dataset structure.

Overall Performance

Across the complete dataset:

Total Amount: 6,182,639
Total Profit: 1,610,697
Total Quantity: 12,745
Overall Profit-to-Amount ratio: approximately 26.05%
Category Performance

Total Amount is distributed relatively evenly between the three categories. Office Supplies generated the highest Total Amount at 2,089,510, followed by Electronics at 2,054,456 and Furniture at 2,038,673.

Furniture recorded the highest number of records (407) but the lowest Total Amount. Its Average Amount per record (5,009.02) was also below Electronics (5,294.99) and Office Supplies (5,236.87), indicating that greater record frequency did not translate into the highest overall Amount.

Profit analysis showed that Office Supplies generated the highest Total Profit at 551,575, followed by Furniture at 540,542 and Electronics at 518,580.

However, Furniture produced the highest category-level profit margin at 26.51%, narrowly ahead of Office Supplies at 26.40%. Electronics had the lowest margin at 25.24%. This demonstrates the distinction between generating the greatest absolute profit and generating the greatest profit relative to Amount.

Sub-Category Performance

At sub-category level, Markers generated the highest Total Amount (627,875), narrowly ahead of Tables (625,177). Printers, however, generated the highest Average Amount per record at 5,961.67, despite having fewer records than several other sub-categories.

Record frequency, Total Amount and Average Amount showed that sub-category performance is influenced by a combination of volume and value per record. Tables had the highest record count (122), whereas Markers combined a relatively high record count (110) with a high Average Amount (5,707.95) to produce the highest Total Amount.

Markers also generated the highest Total Profit at 174,749, followed by Tables (156,796) and Paper (149,723).

Profit-margin analysis provided a different ranking. Bookcases recorded the highest sub-category profit margin at 28.56%, closely followed by Paper (28.53%) and Chairs (28.45%). Phones recorded the lowest margin at 22.58%. This again shows that the products generating the greatest absolute Amount or Profit are not necessarily those with the greatest relative profitability.

Quantity analysis found that Furniture recorded the highest Total Quantity (4,441), followed by Electronics (4,258) and Office Supplies (4,046). However, Electronics had the highest Average Quantity per record (10.97), narrowly ahead of Furniture (10.91).

At sub-category level, Tables recorded the highest Total Quantity (1,303), while Printers had the highest Average Quantity per record (11.83). This provides further evidence that total performance and per-record performance should be considered separately.

State Performance

Geographical analysis found New York generated the highest Total Amount (1,130,048), followed by Florida (1,091,174) and California (1,086,436).

However, Florida generated the highest Total Profit at 308,706, narrowly exceeding New York's 308,506, despite New York having the greater Total Amount.

Profit-margin analysis helps explain this difference. Florida recorded the highest state profit margin at 28.29%, compared with 27.30% for New York. California followed at 25.66%, Texas at 25.49%, Illinois at 24.56% and Ohio at 24.47%.

The state analysis therefore shows that higher Total Amount does not necessarily result in higher Total Profit when profitability differs between locations.

City Performance

City-level analysis was approached by comparing Total Amount, record frequency and Average Amount rather than relying on Total Amount alone.

The five highest cities by Total Amount were Orlando (452,158), San Francisco (440,000), Buffalo (418,514), Rochester (407,291), and Dallas (390,144).

Record counts revealed a different ranking. Buffalo had 90 records, San Francisco 84, Orlando 77 and Rochester 74. Despite Buffalo having 13 more records than Orlando, Orlando generated 33,644 more in Total Amount.

Average Amount helped explain the difference. Orlando recorded the highest Average Amount of all cities at 5,872.18, followed by Miami (5,619.41) and Rochester (5,503.93). Therefore, Orlando's leading Total Amount appears to result from high-value records rather than simply having the greatest record frequency.

City profitability revealed another distinction. Orlando generated the highest Total Profit at 128,125, followed by Buffalo (111,823), Miami (109,739), Rochester (109,729) and San Francisco (108,106).

Miami was particularly notable. Although it did not rank among the five highest cities by Total Amount, it ranked second for Average Amount (5,619.41), second for Average Profit (1,662.71), third for Total Profit (109,739), and first for profit margin (29.59%).

When cities were ranked by profit margin, the top five were Miami (29.59%), Peoria (28.84%), New York City (28.58%), Orlando (28.34%), and Rochester (26.94%). Peoria, New York City and Orlando were particularly close, with only 0.50 percentage points separating their margins.

Overall, the analysis so far consistently shows that high total performance can arise through different combinations of transaction frequency, value per record and profitability. Consequently, Total Amount alone does not provide a complete picture of product or geographical performance.
